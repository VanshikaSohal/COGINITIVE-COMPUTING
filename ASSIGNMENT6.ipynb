```python
#Q1
import numpy as np
import matplotlib.pyplot as plt
M = float(input("Enter the value of M: "))
x = np.linspace(-10, 10, 1000)
y1 = M * x**2
y2 = M * np.sin(x)
plt.plot(x, y1, color="blue", linestyle="-", label="M*x^2")
plt.plot(x, y2, color="red", linestyle="--", label="M*sin(x)")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Mathematical Functions")
plt.legend()
plt.grid(True)
plt.show()
```

```python
#Q2
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
data = {
    "Subject": ["DSA", "DMS", "OS", "NLA", "OOPS"],
    "Score": [85, 78, 92, 80, 88]
}
df = pd.DataFrame(data)
sns.barplot(x="Subject", y="Score", data=df, palette="Set2", hue="Subject", legend=False)
for i in range(len(df)):
    plt.text(i, df["Score"][i] + 1, str(df["Score"][i]),
             ha="center")
plt.title("Subject-wise Scores")
plt.xlabel("Subjects")
plt.ylabel("Scores")
plt.grid(axis="y")
plt.show()
```

```python
#Q3
import numpy as np
import matplotlib.pyplot as plt
rollno = int(input("Enter your roll number: "))
np.random.seed(rollno)
data = np.random.randn(50)
cumulative = np.cumsum(data)
noise = np.random.randn(50)
plt.figure(figsize=(10, 8))
plt.subplot(2, 2, 1)
plt.plot(cumulative)
plt.title("Cumulative Sum")
plt.xlabel("Index")
plt.ylabel("Cumulative Value")
plt.grid(True)
plt.subplot(2, 2, 2)
plt.scatter(range(50), data + noise)
plt.title("Scatter Plot with Random Noise")
plt.xlabel("Index")
plt.ylabel("Value")
plt.grid(True)
plt.subplot(2, 2, 3)
plt.plot(data)
plt.title("Random Dataset")
plt.xlabel("Index")
plt.ylabel("Value")
plt.grid(True)
plt.subplot(2, 2, 4)
plt.hist(data, bins=10)
plt.title("Distribution of Random Data")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
#q4
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
df = pd.read_csv(r"C:\Users\vansh\Downloads\FOLDER\THAPAR\5TH SEMESTER\COGINITIVE\company_sales_data.csv")
sns.set_theme(style="whitegrid")
# Total profit
plt.figure(figsize=(10, 5))
sns.lineplot(
    x="month_number",
    y="total_profit",
    data=df,
    marker="o"
)
plt.title("Total Profit of All Months")
plt.xlabel("Month")
plt.ylabel("Total Profit")
plt.grid(True)
plt.show()
#  All product sales data 
plt.figure(figsize=(12, 6))
products = [
    "facecream",
    "facewash",
    "toothpaste",
    "bathingsoap",
    "shampoo",
    "moisturizer"
]
for product in products:
    sns.lineplot(
        x="month_number",
        y=product,
        data=df,
        marker="o",
        label=product
    )
plt.title("Product Sales Data")
plt.xlabel("Month")
plt.ylabel("Sales Units")
plt.legend()
plt.grid(True)
plt.show()
# Bar chart for all features / attributes
plt.figure(figsize=(14, 7))
df.plot(
    x="month_number",
    kind="bar",
    figsize=(14, 7)
)
plt.title("Bar Chart of All Features")
plt.xlabel("Month")
plt.ylabel("Values")
plt.grid(axis="y")
plt.show()
```
