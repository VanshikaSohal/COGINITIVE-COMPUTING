```python
# Q6
employees = {
    'EmployeeID': [101,102,103,104,105],
    'Name': ['Alice','Bob','Charlie','Diana','Edward'],
    'Department': ['HR','IT','IT','Marketing','Sales'],
    'Age': [29,34,41,28,38],
    'Salary': [50000,70000,65000,55000,60000],
    'YearsOfExperience': [4,8,10,3,12],
    'JoiningDate': ['2020-03-15','2017-07-19','2013-06-01','2021-02-10','2010-11-25'],
    'Gender': ['Female','Male','Male','Female','Male'],
    'Bonus': [5000,7000,6000,4500,5000],
    'Rating': [4.5,4.0,3.8,4.7,3.5]
}
emp = pd.DataFrame(employees)
emp.to_csv('employees.csv', index=False)

# a
print(emp.shape)

# b
print(emp.info())

# c
print(emp.describe())

# d
print(emp.head())
print(emp.tail(3))

# e
print(emp['Salary'].mean())
print(emp['Bonus'].sum())
print(emp['Age'].min())
print(emp['Rating'].max())

# f
print(emp.sort_values('Salary', ascending=False))

# g
def performanceCategory(rating):
    if rating >= 4.5:
        return 'Excellent'
    elif rating >= 4.0:
        return 'Good'
    else:
        return 'Average'

emp['Performance'] = emp['Rating'].apply(performanceCategory)
print(emp)

# h
print(emp.isnull().sum())

# i
emp = emp.rename(columns={'EmployeeID':'ID'})
print(emp)

# j
print(emp[emp['YearsOfExperience'] > 5])
print(emp[emp['Department'] == 'IT'])

# k
emp['Tax'] = emp['Salary'] * 0.10
print(emp)

# l
emp.to_csv('modifiedEmployees.csv', index=False)
```
