```python
#iv
import math
def isprime(n):#using a function
    if n<2:
        return False
    if n==2:
        return True
    if n%2==0:
        return False
    #check odd divisors upto sqrt(n)
    for i in range(3,int(math.sqrt(n))+1,2):
        if n%i==0:
            return False
    return True
prime=()
for i in randomnumber:
    if isprime(i):
       prime+=(i,)#create a new tuple with the prime number added
print(prime)  
```

```python
import random 
from collections import Counter
counted=Counter(randomnumber) 
mostcommonnum,frequency=counted.most_common(1)[0] #Finds the most occuring element with its frequency- most_common is fixed syntax u cannot use it without underscore
print(mostcommonnum) 
print(frequency)
```

```python
#q4
R=[1,0,2,4,1,7,0,3]# u cannot iterate on complete number so u have to make an array for digits of the roll number
#
A=set()
B=set()#this is the way to create set - u cannot use A={} to create set
for i in R:
    A.add(i*7)#for set it is add and not append 
A
for i in R:
    B.add(i*9)
B
```

```python
#vi
C=set()
unionset=A.union(B)
unionset
# or we can use | for finding union 
```

```python
#vii
C=A.intersection(B)
C
```

```python
#viii
Diff=A.difference(B)
print(Diff)
Diff2=B.difference(A)
print(Diff2)
#difference() returns elements only in the first set and not in the second,
#while symmetric_difference() returns elements in either set but not in both (i.e., combines A − B and B − A).
```

```python
#ix
#symmetric difference
# Method 1: Using symmetric_difference() method
symdiff1 = A.symmetric_difference(B)
# Method 2: Using ^ operator
symdiff2 = A ^ B
print(symdiff1)
print(symdiff2)
```

```python
#q5
my_dict = {
    "name": "Vanshika Sohal",
    "roll_no": "YOUR_ROLL_NUMBER",
    "branch": "CSE",
    "age": 19,
    "city": "Una"
}
print("Original dictionary:")
print(my_dict)
# i
my_dict["location"] = my_dict.pop("city")

# ii
my_dict["cgpa"] = 8.7
print(my_dict)

# iii. 
my_dict["age"] += 1
print(my_dict)

# iv. 
dict_pop = my_dict.copy()
dict_del = my_dict.copy()
removed_value = dict_pop.pop("branch")

print("\nUsing pop():")
print(dict_pop)
print("Returned value:", removed_value)
del dict_del["branch"]
print("\nUsing del:")
print(dict_del)
# pop() removes the key and returns its value,
# while del removes the key but does not return its value.

# v. 
print("\nKey-value pairs:")
for key, value in my_dict.items():
    print(f"{key} → {value}")


# vi. 
print("\nChecking for email:")
if "email" in my_dict:
    print("Email:", my_dict["email"])
else:
    print("Email key does not exist.")


# vii.
friend_dict = {
    "name": "Rahul Sharma",
    "roll_no": "FRIEND123",
    "branch": "CSE",
    "age": 20,
    "city": "Chandigarh"
}

merged_dict = {**my_dict, **friend_dict}

print("\nMerged dictionary:")
print(merged_dict)

# When both dictionaries have the same key,
# the value from the second dictionary wins.

# viii. 
string_dict = {
    key: value
    for key, value in my_dict.items()
    if isinstance(value, str)
}
print("\nDictionary containing only string values:")
print(string_dict)
```
