```python
# Q1
import pandas as pd

fixedEntries = [
    {"question": "what is the annual fee", "answer": "The annual fee is Rs 500.",
     "keywords": "fee cost price charge", "category": "billing"},
    {"question": "how to reset password", "answer": "Go to Settings > Reset Password.",
     "keywords": "password reset login", "category": "account"},
    {"question": "what are your working hours", "answer": "We are open 9 AM to 5 PM.",
     "keywords": "hours timing open time", "category": "general"},
    {"question": "how can i pay the fee", "answer": "You can pay via UPI, card, or net banking.",
     "keywords": "pay payment upi fee", "category": "billing"},
]

personalizedEntries = [
    {"question": "how do i get a refund for a cancelled service", "answer": "Refunds are processed within 5-7 business days to your original payment method.",
     "keywords": "refund cancel money", "category": "billing"},
    {"question": "why was i charged twice this month", "answer": "Duplicate charges are usually reversed automatically within 48 hours, or you can contact support.",
     "keywords": "charge duplicate billing", "category": "billing"},
]

faqEntries = fixedEntries + personalizedEntries
df = pd.DataFrame(faqEntries)
print(df)
```

```python
# Q2
def scoreQuery(query, dataframe):
    queryWords = set(query.lower().split())
    scores = []
    for index, row in dataframe.iterrows():
        keywordWords = set(row['keywords'].lower().split())
        questionWords = set(row['question'].lower().split())
        matched = queryWords.intersection(keywordWords.union(questionWords))
        score = len(matched)
        scores.append(score)
    result = dataframe.copy()
    result['score'] = scores
    result = result[result['score'] > 0].sort_values('score', ascending=False)
    return result

print(scoreQuery("how can i pay my fee", df))
```

```python

# Q3
def sameCategory(categoryName, dataframe):
    return dataframe[dataframe['category'] == categoryName]

print(sameCategory("billing", df))
```

```python
# Q6
def scoreQueryWithTies(query, dataframe):
    queryWords = set(query.lower().split())
    scores = []
    for index, row in dataframe.iterrows():
        keywordWords = set(row['keywords'].lower().split())
        questionWords = set(row['question'].lower().split())
        matched = queryWords.intersection(keywordWords.union(questionWords))
        score = len(matched)
        scores.append(score)
    result = dataframe.copy()
    result['score'] = scores
    result = result[result['score'] > 0]
    if result.empty:
        print("No matches found.")
        return result
    maxScore = result['score'].max()
    topMatches = result[result['score'] == maxScore]
    if len(topMatches) > 1:
        print("Tie detected, showing all top matches:")
        print(topMatches)
    else:
        print(topMatches)
    return topMatches

scoreQueryWithTies("fee payment", df)
scoreQueryWithTies("working hours today", df)
```
